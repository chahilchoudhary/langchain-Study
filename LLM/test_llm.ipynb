# Test the LLM connection (IBM ICA base URL)

Checks that the `OPENAI_API_KEY` and `BASE_URL` in `.env` work with LangChain.

```python
import os
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(), override=True)

API_KEY = os.getenv("OPENAI_API_KEY")
BASE_URL = os.getenv("BASE_URL")

# Never print the key itself, only whether it is set
print("API key loaded:", bool(API_KEY))
print("Base URL      :", BASE_URL)
```

## 1. List the models available on this endpoint
Pick a model name from this list and put it in `MODEL` below.

```python
from openai import OpenAI

client = OpenAI(api_key=API_KEY, base_url=BASE_URL)
try:
    for m in client.models.list():
        print(m.id)
except Exception as e:
    print("Could not list models:", type(e).__name__, e)
```

## 2. Call the model through LangChain

```python
from langchain_openai import ChatOpenAI

MODEL = "gpt-5.1"  # must be a name from the list above

llm = ChatOpenAI(model=MODEL, api_key=API_KEY, base_url=BASE_URL)

response = llm.invoke("Say hello in one short sentence.")
print(response.content)
```

## 3. Same call using the Responses API
Some models only work through the Responses API. Here it is with the same `MODEL`:

```python
llm = ChatOpenAI(model=MODEL, api_key=API_KEY, base_url=BASE_URL, use_responses_api=True)
response = llm.invoke("Say hello in one short sentence.")
print(response.text)
```
